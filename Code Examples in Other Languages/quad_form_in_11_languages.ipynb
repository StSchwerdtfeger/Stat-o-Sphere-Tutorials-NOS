# Python Example:

```python
# The math package is part of pythons' basic library, 
# so no extra installation necessary:
import math 

# Function that represents a solution to the quadratic formula:
def quad_form(a, b, c):
    x1 = (-b + math.sqrt(b**2 - 4*a*c)) / (2*a)
    x2 = (-b - math.sqrt(b**2 - 4*a*c)) / (2*a)
    return [x1, x2]

# Print results (including some text in this case):
print("Crossing point(s) with x-axis, where y = 0:", quad_form(2, 4, 2))
# Output in python is:
# Crossing point(s) with x-axis, where y = 0: [-1.0, -1.0]
# Both results are the same, so in this case there is only one point  
# crossing or better: tangently touching the x-axis at y = 0. 
```

### Some basic information on Fortran syntax:

Implicite none is necessary, since it refers to a past convention, where
variable classes are assigned when variables names start with certain letters,
such that the variable with the name like "x_value" would automatically REAL number
and those starting with I, J, K, L, M, N as integers, since those letters are
commonly used as index counters e.g. in a sum operator... It comes from a time
where minimalism was key and punchcards were used as input format...
BEST: Always set implicite none, such that there will be no indirect implicit assignment.
Also, when you got typos, the program will then stop and not assign a new variable
automatically (which makes debugging daunting)...
